```python
# Load a pretrained YOLOv5 architecture (Small version)
model = YOLO("yolov5s.pt") 

# Execute the training pipeline on your tennis ball dataset
results = model.train(
    data="/Users/bibekdulal/Desktop/Tennis_analysis/training/tennis-ball-detection-6/data.yaml",
    epochs=100,      
    imgsz=640,       
    batch=32,        
    device="mps"     
)
```
